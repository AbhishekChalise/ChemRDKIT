```python
aspirin = Chem.MolFromSmiles("CC(=O)Oc1ccccc1C(=O)O")
```

```python
logp_aspirin = Descriptors.MolLogP(aspirin)
print(logp_aspirin)
```

```python
from rdkit.Chem import Lipinski
hbd_lipinski = Lipinski.NumHDonors(aspirin)
hba_lipinski = Lipinski.NumHAcceptors(aspirin)

print(hbd_lipinski)
print(hba_lipinski)
```

```python
def calculate_violations(SmilesFormat: str):
    mol = Chem.MolFromSmiles(SmilesFormat)
    mw = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    hbd_lipinski = Lipinski.NumHDonors(mol)
    hba_lipinski = Lipinski.NumHAcceptors(mol)

    violation = 0
    
    if mw > 500:
        violation = violation + 1

    if logp > 5:
        violation = violation + 1

    if hbd_lipinski > 5:
        violation = violation + 1

    if hba_lipinski > 10:
        violation = violation + 1

    return violation
```

```python
drugs = {
    "Aspirin": "CC(=O)Oc1ccccc1C(=O)O",
    "Wax":     "CCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCCC",
    "Sugar":   "OC[C@H]1O[C@H](O)[C@H](O)[C@@H](O)[C@@H]1O",
}

result = {}

for drug in drugs:
    violation = calculate_violations(drugs[drug])
    result[drug] = violation     

print(result)

for key in result:
    if result[key] > 2:
        print("Fail")
    elif result[key] == 1:
        print("Borderline")
    else:
        print("PassDrug-Like")
```

```python
def count_elements(smiles, element):
    mol = Chem.MolFromSmiles(smiles)
    counter = 0
    for atoms in mol.GetAtoms():
        symbol = atoms.GetSymbol()
        if symbol == element:
            counter = counter + 1

    print(f"This is the counter:{counter}")
```

```python
count_elements("CC(=O)Oc1ccccc1C(=O)O", "O")
count_elements("CC(=O)Oc1ccccc1C(=O)O", "C")
```
